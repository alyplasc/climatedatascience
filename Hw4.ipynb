```python
def k2c(temp):
    """convert from kelvin to celsius """
    temp=temp-273.15
    return temp
```

```python
def c2k(temp):
    """convert from celsius to kelvin"""
    temp=temp-273.15
    return temp
```

```python
def temp_convert(temp,type): 
    """convert from kelvin or celsius to farenheit"""
    if type=='celsius': 
        temp = (temp*9/5)+32
        print(temp, "F")
        
    if type=='kelvin':
        temp = k2c(temp) 
        new_temp= (temp *(9/5))+32
        print(new_temp, "F")
        

```

```python
temp_convert(7,'kelvin')

temp_convert(0, 'celsius')

temp_convert(-20, 'celsius')




```

```python
def f_conversion(temp, newtype):
    if newtype=='celsius':
        temp = (temp*9/5)+32
        print (temp, "C")
    if newtype=='kelvin':
        temp = k2c(temp) 
        new_temp= (temp *(9/5))+32
        print(new_temp, "K")
```

```python
def meter_convert(feet, inches):
    measurement = feet*12 + inches
    return measurement/39.37
```

```python
meter_convert(4,11)
```

```python
meter_convert(5,11)
```

```python
def convert_back(meters): 
    tot_inch = meters*39.37
    feet = tot_inch // 12
    inches = tot_inch % 12
    
    return feet,inches
```

```python
y = meter_convert(5,11)
convert_back(y)
```

```python
x= meter_convert(3,4)
convert_back(x)
```

```python
g= meter_convert(20, 2)
convert_back(g)
```
